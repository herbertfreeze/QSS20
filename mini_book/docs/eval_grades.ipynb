Quizzes and the exam cannot be made up except in documented emergencies; your lowest quiz is dropped automatically. **Websites pushed after the final deadline receive a 0 for the website component.**

## Details

### Quizzes (10% of grade)

Three short, low-stakes quizzes spread across the first half of the term. They check that you are keeping up with the readings and in-class coding, and surface confusions early. **Your lowest quiz score is dropped**, so the 10% reflects your best two.

### Exam 1 — Take-home (5% of grade)

A short take-home exam covering the first unit of the course (data wrangling, cleaning, and descriptive analysis in pandas). You will receive a Jupyter notebook plus a one-page written component, and turn in both an `.ipynb` and a compiled PDF.

### Final project (85% of grade, split across milestones)

The final project is **individual** and is built up across the term in stages so that you get feedback at each step rather than only at the end:

- **Me Search (10%)** — A short proposal where you identify a question that you personally care about and locate a dataset that could plausibly address it. Graded on fit between question, data, and feasibility.
- **NotebookLM literature review (10%)** — A structured literature review using NotebookLM to synthesize prior work on your topic. You will turn in the source set, the synthesized summary, and a short reflection on what NotebookLM got right and wrong.
- **Initial data analysis — Milestone 1 (15%)** — Your first end-to-end pass at the data: loading, cleaning, descriptive statistics, and at least one visualization that speaks to your question.
- **Milestone 2 (5%)** — Public GitHub repo set up with README, organized `code/` and `output/` directories. Lightweight checkpoint to flag direction issues before the final stretch.
- **Final deliverables (45%)** — A 6-page paper (with a required Agentic Analysis section: AI transcripts + coding-session reflection), a complete public GitHub repository, a public website that demos the project, and a 90–120 second in-class lightning talk + 30 second Q&A. See the "Project Components" page for the detailed breakdown.

You are expected to use AI coding assistants for both analysis and website generation. The agentic workflow is part of what we're evaluating — document it in the paper's Agentic Analysis section.
