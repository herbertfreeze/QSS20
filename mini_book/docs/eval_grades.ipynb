# Evaluation and Grading

## Summary and grading policy

Your grade in QSS 20 is built from three things: low-stakes weekly **quizzes**, two short **take-home exams**, and a multi-stage **final project** that you build up across the term (Me Search → NotebookLM literature review → initial data analysis → Milestone 2 → final deliverables). There are no traditional weekly problem sets — the project milestones are where you practice the skills from class on data you actually care about.

The final project is **individual** end-to-end (no team merge step).

**Late policy.** Project milestones may be submitted up to 48 hours late with a 10 percentage point per day deduction. Quizzes and exams cannot be made up except in documented emergencies; your lowest quiz is dropped automatically.

Grades may be curved if no students receive A's without a curve.

## Details

### Quizzes (10% of grade)

Five short, low-stakes quizzes spread across the term, given roughly every other week. They check that you are keeping up with the readings and in-class coding, and surface confusions early. **Your lowest quiz score is dropped**, so the 10% reflects your best four.

### Exam 1 — Take-home (5% of grade)

A short take-home exam covering the first unit of the course (data wrangling, cleaning, and descriptive analysis in pandas). You will receive a Jupyter notebook plus a one-page written component, and turn in both an `.ipynb` and a compiled PDF.

### Exam 2 — Agentic, take-home (10% of grade)

A take-home exam focused on the second half of the course, where the emphasis is on using AI coding assistants effectively and responsibly. You will complete a data task and then reflect on your workflow with the assistant — what you asked, what you accepted, what you rejected, and why. Graded on both the final artifact and the quality of your reflection.

### Final project (75% of grade, split across milestones)

The final project is **individual** and is built up across the term in stages so that you get feedback at each step rather than only at the end. The five milestones are:

- **Me Search (10%)** — A short proposal where you identify a question that you personally care about and locate a dataset that could plausibly address it. Graded on fit between question, data, and feasibility.
- **NotebookLM literature review (10%)** — A structured literature review using NotebookLM to synthesize prior work on your topic. You will turn in the source set, the synthesized summary, and a short reflection on what NotebookLM got right and wrong.
- **Initial data analysis — Milestone 1 (15%)** — Your first end-to-end pass at the data: loading, cleaning, descriptive statistics, and at least one visualization that speaks to your question.
- **Milestone 2 (5%)** — Public GitHub repo with README, organized `code/` and `output/` directories, and real analytical progress. We use this as a checkpoint to flag direction issues before the final stretch.
- **Final deliverables (35%)** — A ~2-minute in-class presentation, a 6-page scientific report, a complete public GitHub repository, and a website that demos the project. See the "Project Components" page for the detailed breakdown.
