# Evaluation and Grading

## Summary and grading policy

Your grade in QSS 20 is built from **quizzes**, a short take-home **Homework 1** early in the term, a multi-stage **final project** that you build up across the term (Me Search → NotebookLM literature review → initial data analysis → Milestone 2 → final deliverables), **in-class participation**, and **DataCamp** modules.

The final project is **individual** end-to-end (no team merge step). You are expected to use AI coding assistants for both analysis and website generation, and to document that workflow in your final paper.

**Late policy.** Project milestones may be submitted up to 48 hours late with a 10 percentage point per day deduction. Quizzes and Homework 1 cannot be made up except in documented emergencies. **Websites pushed after the final deadline receive a 0 for the website component.**

Grades may be curved if no students receive A's without a curve.

## Details

### Quizzes (18%)

Three short, in-class quizzes spread across the first half of the term, each placed after the unit it tests. All three count — no drop.

### Homework 1 (12%)

A short take-home assignment covering the first unit of the course (data wrangling, cleaning, and descriptive analysis in pandas). You will receive a Jupyter notebook plus a one-page written component, and turn in both an `.ipynb` and a compiled PDF.

### Project-related assignments (15%)

- **Me Search (5%)** — A short proposal where you identify a question that you personally care about and locate a dataset that could plausibly address it.
- **NotebookLM literature review (10%)** — A structured literature review using NotebookLM to synthesize prior work on your topic, with verification of NotebookLM's claims against the original sources.

### Final Project (45%)

The final project is **individual** and is built up across the term in stages:

- **Milestone 1 — Initial data analysis (10%)** — Your first end-to-end pass at the data: loading, cleaning, descriptive statistics, and at least one visualization.
- **Milestone 2 — Repo setup (5%)** — Public GitHub repo with README and organized `code/` and `output/` directories.
- **Final package (30%)** — A 6-page paper (with required Agentic Analysis section), a complete public GitHub repository, a public website that demos the project, and a 90–120 second in-class lightning talk. See the "Project Components" page for the breakdown.

You are expected to use AI coding assistants for both analysis and website generation. Document that workflow in the paper's Agentic Analysis section.

### Participation (5%)

In-class workshops and offline engagement (Slack, peer help, attendance, contribution during work sessions).

### DataCamp (5%)

Completion-based. Work through the assigned DataCamp modules on the schedule by the end of the term to receive credit; full credit for completing all of them.
